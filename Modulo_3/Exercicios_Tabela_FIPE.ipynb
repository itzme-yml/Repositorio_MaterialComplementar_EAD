### 🔴 Nível 4: Um pouco de Visualização (Gráficos)

Uma imagem vale mais que mil palavras. Vamos criar gráficos simples.

**Exercício 9:**
Crie um **gráfico de barras** mostrando os 10 veículos com os maiores `valor`.
*Pergunta:* Quais marcas dominam o topo da lista?

*Dica: Use `df.sort_values(by='valor', ascending=False).head(10).plot(kind='bar', x='modelo', y='valor')`*

```python
# Exercício 9: gráfico de barras dos 10 mais caros
top_10 = df.sort_values(by='valor', ascending=False).head(10)
top_10.plot(kind='bar', x='modelo', y='valor', figsize=(12, 6))
plt.title('10 veículos com maior valor')
plt.xlabel('Modelo')
plt.ylabel('Valor (R$)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**Exercício 10:**
Crie um **histograma** para ver como os preços (`valor`) estão distribuídos. A maioria dos veículos é barata ou cara?

*Dica: Use `df['valor'].plot(kind='hist', bins=20)`*

```python
# Exercício 10: histograma dos preços
df['valor'].plot(kind='hist', bins=20, figsize=(10, 5))
plt.title('Distribuição dos preços')
plt.xlabel('Valor')
plt.ylabel('Frequência')
plt.show()
```
